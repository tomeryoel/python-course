```python
def my_func(x,y):
    if x % y == 0:
        return True
    else:
        return False
# my_func(10,5)  # True
print(my_func(10,5))

def my_func(*_noMatter):
    if _noMatter[0] % _noMatter[1] == 0:
        return True
    else:
        return False

#my_func(3,5,7,4,2,8,9,5)  # True
print(my_func(3,5,7,4,2,8,9,5))
```
