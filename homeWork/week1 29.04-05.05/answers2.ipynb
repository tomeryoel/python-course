<h3> QUESTION 3 </h3>

```python
def two_lists(lst1, lst2):
    if len(lst1) != len(lst2):
        return "Error: Lists must be of the same length."
    
    result = []

    for i in range(len(lst1)):
        if lst1[i] > lst2[i]:
            result.append(lst1[i])
        else:
            result.append(lst2[i])

    return print(result)
two_lists([18, 22, 13, 4 ,45], [44, 59, 68 , 17 , 68])
    
```

<h3> QUESTION 4 </h3>

```python
def odd_and_even(lst):

    count_even = 0
    count_odd = 0
    found_string = False
    for item in lst:
        if type(item) == str:
            found_string = True
            break
    
        if item % 2 == 0:
            count_even +=1
        else:
            count_odd +=1 
    if found_string:
            print("It’s a string!!!")
            return
   
    print("Number of even numbers :",  count_even , "Number of odd numbers:", count_odd)
    return
odd_and_even([1, 2, 3, 4, 5, 6, "hello", 7, 8, 9])
odd_and_even([1, 2, 3, 4, 5, 6, 7, 8, 9])
```

<h3> QUESTION 7 </h3>

```python
def count_chars(stl):
    char_count ={}
    for char in stl:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count

print(count_chars("HANNA"))
```

<h3> QUESTION 8 </h3>

```python
def dict_combine(d1, d2):
    dict3 = {}

    for key, val in d1.items():
        if key in d2:
            dict3[key] = d1[key] + d2[key]
        else:
            dict3[key] = val

    for key, val in d2.items():
        if key not in dict3:
            dict3[key] = val

    return dict3
dict_combine({"a": 100, "b": 200, "c": 300}, {"a": 300, "b": 200, "d": 400})
```

<h3> QUESTION 9 </h3>

```python
def return_uniqe(lst):
    new_lst = []
    for num in lst:
        if num not in new_lst:
            new_lst.append(num)
    return new_lst 
print(return_uniqe([1, 2, 3, 3, 3, 3, 4, 5]))
```

```python
def ascending_pattern(n):
    ascending_nums = []
    for num in range(1, n+1):
        ascending_nums.append(num)
        print(*ascending_nums, sep="")
ascending_pattern(8)
```
