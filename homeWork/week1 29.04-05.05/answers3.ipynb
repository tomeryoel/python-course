<h3> QUESTION 1 </h3>

```python
class Vehicle:
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage


vehicle1 = Vehicle("Toyota", 180, 50000)

print(vehicle1.name)
print(vehicle1.max_speed)
print(vehicle1.mileage)
```

```python
class Bus(Vehicle):
    pass

bus1 = Bus("School Bus", 120, 30000)

print(bus1.name)
print(bus1.max_speed)
print(bus1.mileage)
```

<h3> QUESTION 5 </h3>

```python

def count_words(file_path):
    word_count = {}

    with open(file_path, "r") as file:
        text = file.read()

    words = text.split()

    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return word_count


print(count_words("my_id.txt"))

# C:\Users\tomershi\Desktop\python course\homeWork\week1 29.04-05.05\my_id.txt
```

<h3> QUESTION 6 </h3>

```python
def longest_word(file_path):
    with open(file_path, "r") as file:
        text = file.read()

    words = text.split()
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


print(longest_word("my_id.txt"))
```

<h3> QUESTION 7 </h3>

```python
def sum_list(numbers):
    total = 0

    for num in numbers:
        total += num

    return total


print(sum_list([1, 2, 3, 4, 5]))
```
