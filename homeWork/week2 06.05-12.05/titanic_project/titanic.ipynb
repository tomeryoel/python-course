<h3> Loading the Data and analyzing restrictions </h3>

```python
import pandas as pd
import random


df = pd.read_csv('titanic.csv')


df_clean = df.dropna(subset=['fare', 'age', 'survived', 'sex', 'pclass', 'ticket'])


max_fare_total = df_clean['fare'].max()
min_fare_total = df_clean['fare'].min()


class_ranges = {}
for pclass in sorted(df_clean['pclass'].unique()):
    class_df = df_clean[df_clean['pclass'] == pclass]
    class_ranges[pclass] = {
        'min': class_df['fare'].min(),
        'max': class_df['fare'].max()
    }

print(f"Data analysis complete. Fare range: {min_fare_total:.2f}$ - {max_fare_total:.2f}$")
```

<h3> Get the data from the user </h3>

```python

name = input("Please enter your name: ").strip()


while True:
    try:
        age = float(input("Please enter your age (0-130): "))
        if 0 <= age <= 130:
            break
        print("Your age is wrong.")
    except ValueError:
        print("Please enter a valid number for age.")


while True:
    sex = input("Please enter your sex (male/female): ").lower()
    if sex in ['male', 'female']:
        break
    print("Sex value is illegal.")


while True:
    try:
        fare = float(input(f"Enter fare (between {min_fare_total:.2f} and {max_fare_total:.2f}): "))
        if min_fare_total <= fare <= max_fare_total:
            break
        print("Your fare payment is illegal.")
    except ValueError:
        print("Please enter a valid number for fare.")
```

<h3> Calculating death odds and providing a ticket </h3>

```python
import os


assigned_class = 3
for pclass in [1, 2, 3]:
    if class_ranges[pclass]['min'] <= fare <= class_ranges[pclass]['max']:
        assigned_class = pclass
        break


relevant_group = df_clean[(df_clean['sex'] == sex) & (df_clean['pclass'] == assigned_class)]
if len(relevant_group) == 0:
    survival_rate = df_clean['survived'].mean()
else:
    survival_rate = relevant_group['survived'].mean()

death_chance = (1 - survival_rate) * 100


existing_tickets = set(df['ticket'].astype(str).tolist())
while True:
    ticket_num = str(random.randint(100000, 999999))
    if ticket_num not in existing_tickets:
        break


survival_message = f"Dear {name}, your chances to die on our trip are {death_chance:.1f}%.\nEnjoy your trip ☺"
ticket_content = f"""
--- TITANIC PASSENGER TICKET ---
Name: {name}
Age: {age}
Sex: {sex}
Fare: {fare}$
Class: {assigned_class}
Ticket Number: {ticket_num}
-------------------------------
{survival_message}
\n"""


old_content = ""
if os.path.exists('my_ticket.txt'):
    with open('my_ticket.txt', 'r', encoding='utf-8') as f:
        old_content = f.read()


with open('my_ticket.txt', 'w', encoding='utf-8') as f:
    f.write(ticket_content + old_content)

print(f"\n{survival_message}")
print(f"\nSuccess! Your ticket has been generated at the top of 'my_ticket.txt'.")
```
